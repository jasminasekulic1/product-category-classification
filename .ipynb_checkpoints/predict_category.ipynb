```python
import pickle

with open("model.pkl", "rb") as f:
    model = pickle.load(f)

print("Model je učitan.")
print("Unesi naziv proizvoda.")
print("Za kraj upiši: kraj")

while True:
    proizvod = input("Naziv proizvoda: ")

    if proizvod.lower() == "kraj":
        print("Kraj programa.")
        break

    kategorija = model.predict([proizvod])[0]

    print("Predložena kategorija:", kategorija)
```
